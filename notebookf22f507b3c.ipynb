```python
from kaggle_secrets import UserSecretsClient
from google import genai

# Get API key from Kaggle Secrets
secrets = UserSecretsClient()
api_key = secrets.get_secret("GEMINI_API_KEY")

# Connect to Gemini
client = genai.Client(api_key=api_key)

print("Gemini client initialized successfully!")
```

```python
import json
from pathlib import Path

# Search for the uploaded Lumen dataset
files = list(
    Path("/kaggle/input").rglob(
        "Lumen_Task4_Gemini4Neon_Prompts.json"
    )
)

if not files:
    print("Dataset not found!")
    print("Please add your Lumen dataset using Add Input.")
else:
    data_path = files[0]

    with open(data_path, "r", encoding="utf-8") as f:
        dataset = json.load(f)

    cases = dataset["cases"]

    print("Dataset loaded successfully!")
    print("Total examples:", len(cases))
    print("First example:", cases[0]["eval_id"])
    print("Last example:", cases[-1]["eval_id"])
```
